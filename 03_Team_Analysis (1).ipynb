# 03 - Team Analysis

```python
import pandas as pd
import numpy as np

matches = pd.read_csv('data/matches_clean.csv')
deliveries = pd.read_csv('data/deliveries_clean.csv')
print(matches.shape, deliveries.shape)
```

## Matches played & won per team

```python
played = pd.concat([matches['team1'], matches['team2']]).value_counts()
wins = matches['winner'].value_counts().drop('No Result', errors='ignore')

team_stats = pd.DataFrame({'played': played, 'won': wins}).fillna(0)
team_stats['win_pct'] = (team_stats['won'] / team_stats['played'] * 100).round(2)
team_stats.sort_values('win_pct', ascending=False)
```

## Most successful team each season

```python
season_wins = matches.groupby(['season', 'winner']).size().reset_index(name='wins')
season_wins = season_wins[season_wins['winner'] != 'No Result']
season_wins.loc[season_wins.groupby('season')['wins'].idxmax()]
```

## IPL champions (winner of the final)

```python
finals = matches[matches['match_type'] == 'Final']
finals[['season', 'team1', 'team2', 'winner']].sort_values('season')
```

```python
finals['winner'].value_counts()
```

## Toss analysis

```python
matches['toss_win_match_win'] = np.where(matches['toss_winner'] == matches['winner'], 1, 0)
valid = matches[matches['winner'] != 'No Result']
print('Toss winner also won the match: %.2f%%' % (valid['toss_win_match_win'].mean() * 100))
```

```python
print(matches['toss_decision'].value_counts())
print()
print(matches.groupby('season')['toss_decision'].value_counts().unstack().fillna(0))
```

## Home advantage (city vs team)

```python
print(pd.crosstab(valid['city'], valid['winner']).loc[['Mumbai', 'Chennai', 'Kolkata']].T.sort_values('Mumbai', ascending=False).head(8))
```

## Head to head

```python
def head_to_head(a, b):
    m = matches[((matches['team1'] == a) & (matches['team2'] == b)) |
                ((matches['team1'] == b) & (matches['team2'] == a))]
    return m['winner'].value_counts()

head_to_head('Mumbai Indians', 'Chennai Super Kings')
```

## Highest team totals

```python
innings = deliveries.groupby(['match_id', 'inning', 'batting_team'])['total_runs'].sum().reset_index()
innings.sort_values('total_runs', ascending=False).head(10)
```

```python
print('Average 1st innings score:', innings[innings['inning'] == 1]['total_runs'].mean().round(1))
print('Average 2nd innings score:', innings[innings['inning'] == 2]['total_runs'].mean().round(1))
```
